```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('coding/data/customer_events.csv')
print(df.head())
```

```python
df_clean = df.copy()
df_clean['revenue'] = df_clean['price'] * df_clean['quantity']
```

```python
# Flag 1 if event is a purchase, 0 otherwise
df_clean['is_purchase'] = (df_clean['event_type'] == 'purchase').astype(int)

# Flag 1 only if it was a purchase AND it was NOT returned
df_clean['is_non_returned_purchase'] = (
    (df_clean['event_type'] == 'purchase') & (~df_clean['is_returned'])
).astype(int)
```

```python
customer_base = df_clean.groupby('customer_id').agg(
    gross_revenue=('revenue', 'sum'),
    purchase_count=('is_purchase', 'sum'),
    non_returned_purchase_count=('is_non_returned_purchase', 'sum'),
    total_returns=('is_returned', 'sum'),
    unique_products=('product_id', 'nunique'),
    total_actions=('event_type', 'count')
)

print(customer_base.head())
```
