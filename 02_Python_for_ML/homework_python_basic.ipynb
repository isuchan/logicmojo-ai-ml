```python
# Program 1: Swap Two Numbers

def swapTwoNumber(number1, number2):    
    temp = number2
    number2 = number1
    number1 = temp

    return number1, number2

number1 = int(input("Enter first number: "))    
number2 = int(input("Enter second number: "))
swappedNumber1, swappedNumber2 = swapTwoNumber(number1, number2)
print(swappedNumber1, swappedNumber2)
```

```python
# Program 2: Area of a Circle

def areaOfCircle(radius):
    area = 3.14 * radius * radius
    return area

radius = float(input("Enter a radius of circle: "))
area = areaOfCircle(radius)
print(area)
```

```python
# Program 3: Celsius to Fahrenheit

def celsiusToFahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = celsiusToFahrenheit(celsius)
print(fahrenheit)
```

```python
# Program 4: Simple Interest
def simpleInterest(principal, rate, time):
    
    try:
        principal = float(principal)
        rate = float(rate)
        time = float(time)
        
        interest = (principal * rate * time) / 100
        return interest
    except Exception as e:
        print(e)
        return
    
simpleInterestInput = input("Enter principal, rate and time separated by space: ")
inputs = simpleInterestInput.split()
simpleInterest(inputs[0], inputs[1], inputs[2])
```

```python
# Program 5: Even or Odd
def evenOrOdd(number):
    try:
        number = int(number)
        if number % 2 == 0:
            return "Even"
        else:
            return "Odd"
    except Exception as e:
        print(e)
        return
    
number = input("Enter a number: ")
result = evenOrOdd(number)
print(result)
```

```python
# Program 6: Positive / Negative / Zero
def checkNumber(number):
    try:
        number = float(number)
        if number > 0:
            return "Positive"
        elif number < 0:
            return "Negative"
        else:
            return "Zero"
    except Exception as e:
        print(e)
        return

number = input("Enter a number: ")
result = checkNumber(number)
print(result)
```

```python
# Program 7: Largest of Three Numbers

def largestOfThree(number1, number2, number3):
    try:
        number1 = int(number1)
        number2 = int(number2)
        number3 = int(number3)
        
        largest = number1
        if number2 > largest:
            largest = number2
        if number3 > largest:
            largest = number3
        return largest
    except Exception as e:
        print(e)
        return
    
inputs = input("Enter three numbers separated by space: ").split()
largest = largestOfThree(inputs[0], inputs[1], inputs[2])
print(largest)
```

```python
# Program 8: Leap Year Check

def isLeapYear(year):
    try:
        year = int(year)
        if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
            return True
        else:
            return False
    except Exception as e:
        print(e)
        return
    
year = input("Enter a year: ")

if isLeapYear(year):
    print("Leap year")
else:
    print("Not a leap year")
```

```python
# Program 9: Grade from Marks
def gradeByMarks(marks):
    try:
        marks = int(marks)
        if marks >= 90:
            return "A"
        elif marks >= 80:
            return "B"
        elif marks >= 70:
            return "C"
        elif marks >= 60:
            return "D"
        else:
            return "F"
    except Exception as e:
        print(e)
        return
    
marks = input("Enter marks: ")
grade = gradeByMarks(marks)
print(grade)
```

```python
# Program 11: Sum of First N Natural Numbers

def sumNaturalNumbers(n):
    try:
        n = int(n)
        total = 0
        for i in range(1, n + 1):
            total += i
        return total
    except Exception as e:
        print(e)
        return
    
n = input("Enter a number: ")
total = sumNaturalNumbers(n)
print(total)
```

```python
# Program 12: Factorial of N

def generateFactorial(n):
    try:
        n = int(n)
        if n < 0:
            return "Factorial is not defined for negative numbers"
        factorial = 1
        for i in range(1, n + 1):
            factorial *= i
        return factorial
    except Exception as e:
        print(e)
        return
    
n = input("Enter a number: ")
factorial = generateFactorial(n)
print(factorial)
```

```python
# Program 13: Fibonacci (N Terms)

def fibonacciList(n):
    try:
        n = int(n)
        fib_sequence = []
        a, b = 0, 1
        for _ in range(n):
            fib_sequence.append(a)
            a, b = b, a + b
        return fib_sequence
    except Exception as e:
        print(e)
        return
    
n = input("Enter number of terms: ")
fibList = fibonacciList(n)
print(fibList)
```

```python
# Program 14: Prime Check

def isPrime(number):
    try:
        number = int(number)
        if number <= 1:
            return False
        for i in range(2, int(number**0.5) + 1):
            if number % i == 0:
                return False
        return True
    except Exception as e:
        print(e)
        return
    
number = input("Enter a number: ")
if isPrime(number):
    print("Prime")
else:
    print("Not a prime")
```

```python
# Program 16: Sum of Digits

def sumOfDigits(number):
    try:
        number = str(int(number))
        total = 0
        for digit in number:
            total += int(digit)
        return total
    except Exception as e:
        print(e)
        return
    
number = input("Enter a number: ")
digitSum = sumOfDigits(number)
print(digitSum)
```

```python
# Program 18: Palindrome String

def isPalindrome(s):
    s = s.lower().replace(" ", "")  # Normalize the string
    return s == s[::-1]

string = input("Enter a string: ")
if isPalindrome(string):
    print("Palindrome")
```

```python
# Program 19: Count Vowels

def countVowels(s):
    s = s.lower()
    vowels = 'aeiou'
    count = 0
    for char in s:
        if char in vowels:
            count += 1
    return count

string = input("Enter a string: ")
vowelCount = countVowels(string)
print(vowelCount)
```

```python
# Program 22: Sum of List Elements

def sumOfListElements(lst):
    total = 0
    for element in lst:
        total += element
    return total

inputLength = input("Enter number of elements in the list: ")
inputs = input("Enter numbers separated by space: ").split()

if(len(inputs) != int(inputLength)):
    print("Number of elements does not match the specified length.")
else:
    numbers = [int(num) for num in inputs]
    totalSum = sumOfListElements(numbers)
    print(totalSum)
```

```python
# Program 23: Largest and Smallest in a List

def largestAndSmallest(lst):
    if not lst:
        return None, None
    largest = lst[0]
    smallest = lst[0]
    for element in lst:
        if element > largest:
            largest = element
        if element < smallest:
            smallest = element
    return largest, smallest

inputLength = input("Enter number of elements in the list: ")
inputs = input("Enter numbers separated by space: ").split()

if(len(inputs) != int(inputLength)):
    print("Number of elements does not match the specified length.")
else:
    numbers = [int(num) for num in inputs]
    largest, smallest = largestAndSmallest(numbers)
    print(f"Largest: {largest}, Smallest: {smallest}")
```

```python
# Program 24: Count Occurrences in a List

def countOccurrences(lst, target):
    count = 0
    for element in lst:
        if element == target:
            count += 1
    return count

inputLength = input("Enter number of elements in the list: ")
inputs = input("Enter numbers separated by space: ").split()
target = int(input("Enter the number to count occurrences of: "))

if(len(inputs) != int(inputLength)):
    print("Number of elements does not match the specified length.")
else:
    numbers = [int(num) for num in inputs]
    occurrenceCount = countOccurrences(numbers, target)
    print(occurrenceCount)
```

```python
# Program 25: Even Numbers from a List

def evenNumbersFromList(lst):
    evenNumbers = []
    for element in lst:
        if element % 2 == 0:
            evenNumbers.append(element)
    return evenNumbers
inputLength = input("Enter number of elements in the list: ")
inputs = input("Enter numbers separated by space: ").split()

if(len(inputs) != int(inputLength)):
    print("Number of elements does not match the specified length.")
else:
    numbers = [int(num) for num in inputs]
    evenNumbers = evenNumbersFromList(numbers)
    print(evenNumbers) 
```

```python
# Program 26: Reverse a List (without reverse())

def reverseList(lst):
    reversedList = []
    for i in range(len(lst) - 1, -1, -1):
        reversedList.append(lst[i])
    return reversedList

inputLength = input("Enter number of elements in the list: ")
inputs = input("Enter numbers separated by space: ").split()

if(len(inputs) != int(inputLength)):
    print("Number of elements does not match the specified length.")
else:
    numbers = [int(num) for num in inputs]
    reversedNumbers = reverseList(numbers)
    print(reversedNumbers)
```

```python
# Program 27: Remove Duplicates (Preserve Order)

def removeDuplicates(lst):
    seen = set()
    uniqueList = []
    for element in lst:
        if element not in seen:
            seen.add(element)
            uniqueList.append(element)
    return uniqueList

inputLength = input("Enter number of elements in the list: ")
inputs = input("Enter numbers separated by space: ").split()

if(len(inputs) != int(inputLength)):
    print("Number of elements does not match the specified length.")
else:
    numbers = [int(num) for num in inputs]
    uniqueNumbers = removeDuplicates(numbers)
    print(uniqueNumbers)
```

```python
# Program 28: Second Largest (Distinct)

def second_largest(lst):
    distinctList = list(set(lst))
    if len(distinctList) < 2:
        return None
    distinctList.sort()
    return distinctList[-2]

inputLength = input("Enter number of elements in the list: ")
inputs = input("Enter numbers separated by space: ").split()

if(len(inputs) != int(inputLength)):
    print("Number of elements does not match the specified length.")
else:
    numbers = [int(num) for num in inputs]
    secondLargestNumber = second_largest(numbers)
    print(secondLargestNumber)
```

```python
# Program 29: Rotate List Right by K

def rotateListRightByIndex(lst, k):
    n = len(lst)
    k = k % n  # Handle cases where k > n
    return lst[-k:] + lst[:-k]

inputLength = input("Enter numbers separated by space: ")
inputs = input("Enter numbers separated by space: ").split()
rotationIndex = int(input("Enter number of positions to rotate: "))

if(len(inputs) != int(inputLength)):
    print("Number of elements does not match the specified length.")
else:
    numbers = [int(num) for num in inputs]

    rotatedList = rotateListRightByIndex(numbers, rotationIndex)
    print(rotatedList)
```
