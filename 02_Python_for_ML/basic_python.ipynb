```python
#shows the line of code execution within try catch block
a = 10
print("Example integer:", a, type(a))
try:
    print("Example integer:", a, type(a), float(a), isinstance(a, int))
    print("Example string:", a, type(a), int(a), isinstance(a, string))
except Exception as e:
    print("Error:", e)
print("line ended")    
```

```python
#Arithmetic Operations
import math

x = 15
y = 4
print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Modulus:", x % y)
print("Exponentiation:", x ** y)
print("Floor Division:", x // y) #returns only integer valu
print("Floor Division:", math.floor(2.7)) # returns 2    
print("Ceiling:", math.ceil(2.3)) # returns 3
```

```python
a=[1,2,3]
b=[1,2,3]
a is b #false, because a and b are two different variables which means two different objects in memory
```

```python
a=[1,2,3]
b=a.copy()

a is b #false, because a and b are still two different variables which means two different objects in memory
```

```python
#how to create two list with same memory address so when they are changed both will be changed
a=[1,2,3]
b = a #if this changed into b=a.copy() then both list will not be changed together
print("Before changing b:", a, b)
b[0] = 4
print("After changing b:", a, b)
```

```python
#before f string printing
name="test"
age=25
print("My name is " + name + " and I am " + str(age) + " years old.")

#printing with .format()
print("My name is {} and I am {} years old.".format(name, age))

#NOW printing with f string - preffrered way
print(f"My name is {name} and I am {age} years old.")
```

# LIST

```python
list1 = [10, 20, 30]
print(list1[1:3], list1[:2])
```

```python
#reverse list
list1 = [10,20,30,40,50]
print(list1[::-1])
```

```python
#append vs append
list1 = [1,2]
list2 = [3,4]
list1.append(list2)
print(list1)

list1 = [1,2]
list2 = [3,4]
list1.extend(list2)
print(list1)
```

```python
#insert
list1 = [1,2,5]
list1.insert(2)
print(list1)
```
