```python
## Question: Build a vocabulary set of unique words from these documents

print("=== Example 1: Building Vocabulary ===")
documents = [
    "machine learning is awesome",
    "deep learning is powerful",
    "machine learning and deep learning"
]

dictionarySet = set()
for doc in documents:
    words = doc.lower().split() #add lower here to ensure Awesome is equal to awesome
    for word in words:
        dictionarySet.add(word)
        
print("Vocabulary Set:", dictionarySet)

dictionarySet.add("awesome")
print("Updated Vocabulary Set:", dictionarySet)
```

```python
#1: Word Frequency Count
text = "machine learning is great machine learning is powerful learning is fun"

wordCount = {}
words = text.lower().split()
for word in words:
    if word in wordCount:
        wordCount[word] += 1
    else:
        wordCount[word] = 1
print("Word Frequency Count:", wordCount)
```
