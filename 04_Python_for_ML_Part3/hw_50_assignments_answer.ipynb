```python
# Problem 2: Coordinate Distance Calculator
import math
def distance(point1, point2):
    x1, y1 = point1
    x2, y2 = point2
    dist = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
    return dist

distance = distance((1, 2), (4, 6))
print(distance)
```

```python
# Problem 3: Training Configuration Validator

def is_power_of_two(n):
    if n <= 0:
        return False
    return (n & (n - 1)) == 0

def validate_config(config):   
    learning_rate, batch_size, epochs, optimizer = config
    error_messages = []

    if learning_rate < 0.0001 and learning_rate > 1.0:        
        error_messages.append("Learning rate must be between 0.0001 and 1.0")
        
    if(not is_power_of_two(batch_size)):
        error_messages.append("Batch size must be a power of two")
        
    if(epochs < 0):
        error_messages.append("Epochs must be a positive integer")
    
    if (optimizer not in ["sgd", "adam", "rmsprop"]):
        error_messages.append("Optimizer must be one of 'sgd', 'adam', or 'rmsprop'")
    
    is_valid = True if len(error_messages) == 0 else False
    return (is_valid, error_messages)

validate_config = validate_config((0.001, 32, 100, 'sgd'))
print(validate_config)
```

```python
# Problem 4: Data Split Generator

def split_data(total_samples, train_ratio, val_ratio, test_ratio):
    total = int(total_samples)
    train_ratio = int(total * train_ratio)
    val_ratio = train_ratio + int(total * val_ratio)
    test_ratio = val_ratio + int(total * test_ratio)
    
    train_indices = (0, train_ratio)    
    val_indices = (train_ratio, val_ratio)
    test_indices = (val_ratio, total)

    return (train_indices, val_indices, test_indices)


split_data = split_data(1000, 0.8, 0.1, 0.1)
print(split_data)
```

```python
#Problem 5: Model Metrics Comparison

def compare_models(models):    
    output = {
        'best_models': {
            'accuracy': None,
            'precision': None,
            'recall': None,
            'f1_score': None
        },
        'average':{
            'accuracy': None,
            'precision': None,
            'recall': None,
            'f1_score': None
        },
        'ranked_by_f1': []
    }
    
    output['best_models']['accuracy'] = max(models, key=lambda x: x[1])[0]    
    output['best_models']['precision'] = max(models, key=lambda x: x[2])[0]    
    output['best_models']['recall'] = max(models, key=lambda x: x[3])[0]    
    output['best_models']['f1_score'] = max(models, key=lambda x: x[4])[0]
    
    accuracy_values = list(map(lambda x: x[1], models))
    output['average']['accuracy'] = round(sum(accuracy_values) / len(accuracy_values), 2)
    
    precision_values = list(map(lambda x: x[2], models))
    output['average']['precision'] = round(sum(precision_values) / len(precision_values), 2)
    
    recall_values = list(map(lambda x: x[3], models))
    output['average']['recall'] = round(sum(recall_values) / len(recall_values), 2)
    
    f1_score_values = list(map(lambda x: x[4], models))
    output['average']['f1_score'] = round(sum(f1_score_values) / len(f1_score_values), 2)
    
    f1_ranked_models = sorted(models, key=lambda x: x[4], reverse=True)
    for model in f1_ranked_models:
        model_info = model[0]
        output['ranked_by_f1'].append(model_info)   
        
    return output

models = [
    ("Model_A", 0.85, 0.82, 0.88, 0.85),
    ("Model_B", 0.92, 0.90, 0.85, 0.87),
    ("Model_C", 0.88, 0.85, 0.92, 0.88)
]

compare_models(models) 
```

```python
# Problem 6: Duplicate Detector

def find_duplicates(items):
    seen = set()
    duplicates = set()
    output = {
        'has_duplicates': False,
        'unique_count': 0,        
        'duplicate_count': 0,
        'duplicates': {}
    }
    
    for item in items:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
    
    output['has_duplicates'] = True if len(duplicates) > 0 else False
    output['unique_count'] = len(seen)
    output['duplicate_count'] = len(duplicates)
    output['duplicates'] = set(duplicates)
    
    return output

find_duplicates = find_duplicates([1, 2, 3, 2, 4, 3, 5])
print(find_duplicates)
```

```python
# Problem 7: Feature Engineering - Categorical Encoder

def encode_categories(categories):  
    unique_categories = list(set(categories))
    category_to_index = {category: index for index, category in enumerate(unique_categories)}
    output = {
        'encoding_map': {},
        'encoded_data': [],
        'num_categories': 0
    }
    
    for unique_category in unique_categories:
        output['encoding_map'][unique_category] = category_to_index[unique_category]
    
    for category in categories:       
        if category in category_to_index:
            output['encoded_data'].append(category_to_index[category])
    
    output['num_categories'] = len(unique_categories)
    
    return output

encode_categories = encode_categories(['cat', 'dog', 'cat', 'bird', 'cat', 'dog'])
print(encode_categories)
```

```python
# Problem 8: Dataset Overlap Analyzer

def analyze_overlap(train_ids, test_ids):
    train_set = set(train_ids)
    test_set = set(test_ids)
    
    overlapping_ids = train_set.intersection(test_set)
    
    output = {
        'total_unique': 0,
        'overlap': [],
        'only_train': [],
        'only_test': [],
        'has_leakage': False
        
    }
    
    if len(overlapping_ids) > 0:
        output['has_leakage'] = True
        output['overlap'] = list(overlapping_ids)
    
    output['total_unique'] = len(train_set.union(test_set))
    output['only_train'] = list(train_set.difference(test_set))
    output['only_test'] = list(test_set.difference(train_set))
    
    return output

analyze_overlap = analyze_overlap([1, 2, 3, 4, 5], [4, 5, 6, 7, 8])
print(analyze_overlap)
```

```python
# Problem 9: Text Preprocessing - Stopword Remover

def remove_stopwords(text, stopwords):
    words = text.lower().split()
    filtered_words = [word for word in words if word not in stopwords]        
    unique_filtered_words = set()
    
    for word in filtered_words:
        if word not in unique_filtered_words:            
            unique_filtered_words.add(word)
    
    output = {
        'original_text': list(words),
        'filtered_words': filtered_words,
        'num_removed': len(words) - len(filtered_words),
        'unique_filtered': unique_filtered_words
    }
      
    return output

stopwords = {'the', 'is', 'a', 'an', 'in', 'on', 'at'}
text = "the cat is on the mat mat in the house is happy"
print(remove_stopwords(text, stopwords))
```

```python
# Problem 10: Feature Selection - Correlation Filter

def filter_correlated(features, threshold):
    output = {
        'highly_correlated_pairs': set(),
        'features_to_remove': {},
        'featured_to_keep': {}
    }
    
    correlated_features = []
    seen_features = []
    duplicate_features = set()
    
    for i, (key, value) in enumerate(features.items()):
        if(abs(value) >= threshold):
            output['highly_correlated_pairs'].add(key)
            correlated_features.append(key[0])
            correlated_features.append(key[1])
            
    for pair in correlated_features:
        if pair in seen_features:
            duplicate_features.add(pair)
        else:
            seen_features.append(pair)
    
    output['features_to_remove'] = set(duplicate_features)
    output['featured_to_keep'] = set(seen_features) - set(duplicate_features)
        
    return output

correlations = {
    ('f1', 'f2'): 0.95,
    ('f1', 'f3'): 0.5,
    ('f2', 'f4'): 0.92,
    ('f3', 'f4'): 0.3
}

filter_correlated(correlations, threshold=0.9)
```

```python
# Problem 13: Data Aggregator

def aggregate_data(data):
    output = {}
    
    merged_categories = []
    
    for record in data:
        if record['category'] not in merged_categories:
            merged_categories.append(record['category'])
            output[record['category']] = [record['value']]
        else:
            output[record['category']].append(record['value'])
     
    for data in output.items():
        output[data[0]] = {
            'sum': sum(data[1]),
            'avg': round(sum(data[1]) / len(data[1]), 2),
            'count': len(data[1])
        }
             
    
    return output   

data = [
    {'category': 'A', 'value': 10},
    {'category': 'B', 'value': 20},
    {'category': 'A', 'value': 15},
    {'category': 'B', 'value': 25},
    {'category': 'A', 'value': 12}
]
print(aggregate_data(data))
```

```python
# Problem 14: Confusion Matrix Builder

def build_confusion_matrix(actual, predicted):
    output = {
        'matrix': {},
        'accuracy': 0.0,
        'per_class': {}
    }
    
    unique_labels = sorted(set(predicted + actual))
    
    for label in unique_labels:
        output['matrix'][label] = {l: 0 for l in unique_labels}
    
    for a, p in zip(actual, predicted):
        output['matrix'][a][p] += 1
        if(a == p):
            output['accuracy'] += 1
        
    output['accuracy'] /= len(actual)
    
    for label in unique_labels:
        output['per_class'][label] = {
            'precision': round(output['matrix'][label][label] / sum(output['matrix'][l][label] for l in unique_labels) if sum(output['matrix'][l][label] for l in unique_labels) > 0 else 0, 2),
            'recall': round(output['matrix'][label][label] / sum(output['matrix'][label].values()) if sum(output['matrix'][label].values()) > 0 else 0, 2),
            'f1': round(2 * (output['matrix'][label][label] / sum(output['matrix'][l][label] for l in unique_labels) * output['matrix'][label][label] / sum(output['matrix'][label].values())) / (output['matrix'][label][label] / sum(output['matrix'][l][label] for l in unique_labels) + output['matrix'][label][label] / sum(output['matrix'][label].values())) if (output['matrix'][label][label] / sum(output['matrix'][l][label] for l in unique_labels) + output['matrix'][label][label] / sum(output['matrix'][label].values())) > 0 else 0, 2)
        }
    
    return output

actual = ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'C']
predicted = ['A', 'B', 'B', 'C', 'A', 'A', 'C', 'B']
build_confusion_matrix(actual, predicted)
```

```python
# Problem 15: Configuration Merger

def merge_configs(configs):
    output = {}
    
    for config in configs:
        for key, value in config.items():
            if key not in output:
                output[key] = value
            else:
                if isinstance(value, dict) and isinstance(output[key], dict):
                    output[key] = merge_configs([output[key], value])
                else:
                    output[key] = value
    
    return output

configs = [
    {'model': {'layers': 3, 'dropout': 0.5}, 'lr': 0.01},
    {'model': {'layers': 5}, 'batch_size': 32},
    {'model': {'dropout': 0.3}, 'lr': 0.001}
]

merge_configs(configs)
```

```python
# Problem 16: Decorator - Execution Timer
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"slow_function executed in {end_time - start_time:.2f} ms")
        return result
    return wrapper

@timer
def slow_function(n):
    time.sleep(n)
    return n * 2

result = slow_function(0.5)
print(f"Result: {result}")
```

```python
# Problem 17: Memoization Cache

from functools import lru_cache

def memoize(func):
    cache = {}
    stats = {'hits': 0, 'misses': 0}
    def wrapper(*args):
        if args in cache:
            stats['hits'] += 1  
            return cache[args]
        
        stats['misses'] += 1
        result = func(*args)
        
        cache[args] = result        
        return result
    
    def cache_info():
        return {**stats, 'size': len(cache)}
    
    wrapper.cache_info = cache_info
    return wrapper

@memoize
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print("First call:", fibonacci(10))
print("Second call:", fibonacci(10))
print("Cache info:", fibonacci.cache_info())
```

```python
# Problem 18: Partial Function Application

def create_metric(metric_name):
    def metric_function(y_true, y_pred):
        if metric_name == 'accuracy':
            correct = sum(1 for yt, yp in zip(y_true, y_pred) if yt == yp)
            return correct / len(y_true)
        elif metric_name == 'precision':
            true_positive = sum(1 for yt, yp in zip(y_true, y_pred) if yt == yp == 1)
            predicted_positive = sum(1 for yp in y_pred if yp == 1)
            return true_positive / predicted_positive if predicted_positive > 0 else 0
        elif metric_name == 'recall':
            true_positive = sum(1 for yt, yp in zip(y_true, y_pred) if yt == yp == 1)
            actual_positive = sum(1 for yt in y_true if yt == 1)
            return true_positive / actual_positive if actual_positive > 0 else 0
        elif metric_name == 'f1_score':
            precision = metric_function('precision')(y_true, y_pred)
            recall = metric_function('recall')
            return 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
        else:
            raise ValueError("Unsupported metric")
    
    return metric_function

accuracy_fn = create_metric('accuracy')
precision_fn = create_metric('precision')

y_true = [1, 0, 1, 1, 0]
y_pred = [1, 0, 1, 0, 0]

print("Accuracy:", accuracy_fn(y_true, y_pred))
print("Precision:", precision_fn(y_true, y_pred))
```

```python
# Problem 19: Pipeline Builder

def create_pipeline(*functions):
    def pipeline(input_data):
        data = input_data
        for func in functions:
            data = func(data)
        return data
    return pipeline

def normalize(x):
    return [i / max(x) for i in x]

def square(x):
    return [i ** 2 for i in x]

def sum_all(x):
    return sum(x)

pipeline = create_pipeline(normalize, square, sum_all)
result = pipeline([1, 2, 3, 4, 5])
print(f"Result: {result}")
```

```python
# Problem 20: Currying

import inspect

def curry(func):
    def curried_function(*args):
        if len(args) >= len(inspect.signature(func).parameters):
            return func(*args)
        return lambda *more_args: curried_function(*(args + more_args))
    return curried_function

def add_three(a, b, c):
    return a + b + c

curried_add = curry(add_three)
result1 = curried_add(1)(2)(3)
print(f"Curried: {result1}")

add_1 = curried_add(1)
add_1_2 = add_1(2)
result2 = add_1_2(3)
print(f"Partial application: {result2}")
```

```python
#Problem 22: Feature scaler
import statistics

def scale_features(data, method):
    if method == "zscore":
        mean = statistics.mean(data)
        stdev = statistics.stdev(data)
        return [(x - mean) / stdev for x in data]
    elif method == "minmax":
        min_val = min(data)
        max_val = max(data)
        return [(x - min_val) / (max_val - min_val) for x in data]
    elif method == "robust":
        median = statistics.median(data)
        q1 = statistics.quantiles(data, n=4)[0]
        q3 = statistics.quantiles(data, n=4)[2]
        iqr = q3 - q1
        return [(x - median) / iqr for x in data]
    
data = [1, 2, 3, 4, 5]
print("Min-Max", scale_features(data, 'minmax'))
print("Z-score:", scale_features(data, 'zscore'))
print("Robust:", scale_features(data, 'robust'))
```

```python
# Problem 23: reduce() - Custom Aggregations

from functools import reduce

def product(numbers):
    # Multiply x * y cumulatively. Initializer=1 (identity for multiplication)
    return reduce(lambda x, y: x * y, numbers, 1)

def maximum(numbers):
    # Determine max between current accum and next item
    return reduce(lambda x, y: x if x > y else y, numbers)

def join_strings(strings, separator):
    # Concatenate string with separator
    return reduce(lambda x, y: x + separator + y, strings)

def flatten(nested_list):
    # Concatenate lists. Initializer=[]
    return reduce(lambda x, y: x + y, nested_list, [])

print("Product:", product([1, 2, 3, 4, 5]))
print("Maximum:", maximum([3, 1, 4, 1, 5, 9, 2, 6]))
print("Join:", join_strings(['Hello', 'World', 'ML'], ' '))
print("Flatten:", flatten([[1, 2], [3, 4], [5, 6]]))
```

```python
#  Problem 24: Comprehension Olympics

def transpose(matrix):
    # Swap rows and cols: row i col j becomes row j col i
    return [[row[i] for row in matrix] for i in range(len(matrix[0]))]

def flatten_even(matrix):
    # Double loop comprehension with condition
    return [num for row in matrix for num in row if num % 2 == 0]

def cartesian(list1, list2, threshold=5):
    # Double loop with filtering condition
    return [(x, y) for x in list1 for y in list2 if x + y > threshold]

def invert_dict(d):
    # Invert mapping: value becomes key.
    # Since values might not be unique, we collect keys in a list
    unique_vals = set(d.values())
    return {v: [k for k, val in d.items() if val == v] for v in unique_vals}

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Transpose:", transpose(matrix))
print("Flatten even:", flatten_even(matrix))
print("Cartesian:", cartesian([1, 2, 3], [4, 5, 6]))
print("Invert:", invert_dict({'a': 1, 'b': 2, 'c': 1}))
```

```python
# Problem 25: function Combinator
def compose(*functions):
    # Right-to-left execution: f(g(x))
    return lambda x: reduce(lambda acc, f: f(acc), reversed(functions), x)

def pipe(*functions):
    # Left-to-right execution: f(x) -> g(result)
    return lambda x: reduce(lambda acc, f: f(acc), functions, x)

def parallel(*functions):
    # Execute all functions on same input and return list of results
    return lambda x: [f(x) for f in functions]

def conditional(predicate, true_func, false_func):
    # If p(x) is true, run true_func(x), else false_func(x)
    return lambda x: true_func(x) if predicate(x) else false_func(x)

add_2 = lambda x: x + 2
mult_3 = lambda x: x * 3
square = lambda x: x ** 2

print("Compose:", compose(square, mult_3, add_2)(5))
print("Pipe:", pipe(add_2, mult_3, square)(5))
print("Parallel:", parallel(add_2, mult_3, square)(5))
print("Conditional:", conditional(lambda x: x > 0, add_2, mult_3)(5))
print("Conditional:", conditional(lambda x: x > 0, add_2, mult_3)(-5))
```
