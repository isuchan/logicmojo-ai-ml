```python
# Problem 1: Image Dimensions Handler
def image_info(image):
    height, width, channels = image
    total_pixels = height * width
    is_grayscale = True if channels == 1 else False
    return (height, width, channels, total_pixels, is_grayscale)

image_info = image_info((224, 224, 1))
print(image_info)
```

```python
# Problem 2: Coordinate Distance Calculator
import math
def distance(point1, point2):
    x1, y1 = point1
    x2, y2 = point2
    dist = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
    return dist

distance = distance((1, 2), (4, 6))
print(distance)
```

```python
# Problem 3: Training Configuration Validator

def is_power_of_two(n):
    if n <= 0:
        return False
    return (n & (n - 1)) == 0

def validate_config(config):   
    learning_rate, batch_size, epochs, optimizer = config
    error_messages = []

    if learning_rate < 0.0001 and learning_rate > 1.0:        
        error_messages.append("Learning rate must be between 0.0001 and 1.0")
        
    if(not is_power_of_two(batch_size)):
        error_messages.append("Batch size must be a power of two")
        
    if(epochs < 0):
        error_messages.append("Epochs must be a positive integer")
    
    if (optimizer not in ["sgd", "adam", "rmsprop"]):
        error_messages.append("Optimizer must be one of 'sgd', 'adam', or 'rmsprop'")
    
    is_valid = True if len(error_messages) == 0 else False
    return (is_valid, error_messages)

validate_config = validate_config((0.001, 32, 100, 'sgd'))
print(validate_config)
```

```python
# Problem 4: Data Split Generator

def split_data(total_samples, train_ratio, val_ratio, test_ratio):
    total = int(total_samples)
    train_end = int(total * train_ratio)
    val_end = train_end + int(total * val_ratio)
    
    data = list(range(total))
    train_indices = data[:train_end]
    val_indices = data[train_end:val_end]
    test_indices = data[val_end:]
    
    return (train_indices, val_indices, test_indices)

split_data = split_data(1000, 0.7, 0.15, 0.15)
print(split_data)
```
