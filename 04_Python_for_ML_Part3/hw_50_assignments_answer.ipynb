```python
# Problem 1: Image Dimensions Handler
def image_info(image):
    height, width, channels = image
    total_pixels = height * width
    is_grayscale = True if channels == 1 else False
    return (height, width, channels, total_pixels, is_grayscale)

image_info = image_info((224, 224, 1))
print(image_info)
```

```python
# Problem 2: Coordinate Distance Calculator
import math
def distance(point1, point2):
    x1, y1 = point1
    x2, y2 = point2
    dist = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
    return dist

distance = distance((1, 2), (4, 6))
print(distance)
```

```python
# Problem 3: Training Configuration Validator

def is_power_of_two(n):
    if n <= 0:
        return False
    return (n & (n - 1)) == 0

def validate_config(config):   
    learning_rate, batch_size, epochs, optimizer = config
    error_messages = []

    if learning_rate < 0.0001 and learning_rate > 1.0:        
        error_messages.append("Learning rate must be between 0.0001 and 1.0")
        
    if(not is_power_of_two(batch_size)):
        error_messages.append("Batch size must be a power of two")
        
    if(epochs < 0):
        error_messages.append("Epochs must be a positive integer")
    
    if (optimizer not in ["sgd", "adam", "rmsprop"]):
        error_messages.append("Optimizer must be one of 'sgd', 'adam', or 'rmsprop'")
    
    is_valid = True if len(error_messages) == 0 else False
    return (is_valid, error_messages)

validate_config = validate_config((0.001, 32, 100, 'sgd'))
print(validate_config)
```

```python
# Problem 4: Data Split Generator

def split_data(total_samples, train_ratio, val_ratio, test_ratio):
    total = int(total_samples)
    train_ratio = int(total * train_ratio)
    val_ratio = train_ratio + int(total * val_ratio)
    test_ratio = val_ratio + int(total * test_ratio)
    
    train_indices = (0, train_ratio)    
    val_indices = (train_ratio, val_ratio)
    test_indices = (val_ratio, total)

    return (train_indices, val_indices, test_indices)


split_data = split_data(1000, 0.8, 0.1, 0.1)
print(split_data)
```

```python
#Problem 5: Model Metrics Comparison

def compare_models(models):    
    output = {
        'best_models': {
            'accuracy': None,
            'precision': None,
            'recall': None,
            'f1_score': None
        },
        'average':{
            'accuracy': None,
            'precision': None,
            'recall': None,
            'f1_score': None
        },
        'ranked_by_f1': []
    }
    
    output['best_models']['accuracy'] = max(models, key=lambda x: x[1])[0]    
    output['best_models']['precision'] = max(models, key=lambda x: x[2])[0]    
    output['best_models']['recall'] = max(models, key=lambda x: x[3])[0]    
    output['best_models']['f1_score'] = max(models, key=lambda x: x[4])[0]
    
    accuracy_values = list(map(lambda x: x[1], models))
    output['average']['accuracy'] = round(sum(accuracy_values) / len(accuracy_values), 2)
    
    precision_values = list(map(lambda x: x[2], models))
    output['average']['precision'] = round(sum(precision_values) / len(precision_values), 2)
    
    recall_values = list(map(lambda x: x[3], models))
    output['average']['recall'] = round(sum(recall_values) / len(recall_values), 2)
    
    f1_score_values = list(map(lambda x: x[4], models))
    output['average']['f1_score'] = round(sum(f1_score_values) / len(f1_score_values), 2)
    
    f1_ranked_models = sorted(models, key=lambda x: x[4], reverse=True)
    for model in f1_ranked_models:
        model_info = model[0]
        output['ranked_by_f1'].append(model_info)   
        
    return output

models = [
    ("Model_A", 0.85, 0.82, 0.88, 0.85),
    ("Model_B", 0.92, 0.90, 0.85, 0.87),
    ("Model_C", 0.88, 0.85, 0.92, 0.88)
]

compare_models(models) 
```

```python
# Problem 6: Duplicate Detector

def find_duplicates(items):
    seen = set()
    duplicates = set()
    output = {
        'has_duplicates': False,
        'unique_count': 0,        
        'duplicate_count': 0,
        'duplicates': {}
    }
    
    for item in items:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
    
    output['has_duplicates'] = True if len(duplicates) > 0 else False
    output['unique_count'] = len(seen)
    output['duplicate_count'] = len(duplicates)
    output['duplicates'] = set(duplicates)
    
    return output

find_duplicates = find_duplicates([1, 2, 3, 2, 4, 3, 5])
print(find_duplicates)
```

```python
# Problem 7: Feature Engineering - Categorical Encoder

def encode_categories(categories):  
    unique_categories = list(set(categories))
    category_to_index = {category: index for index, category in enumerate(unique_categories)}
    output = {
        'encoding_map': {},
        'encoded_data': [],
        'num_categories': 0
    }
    
    for unique_category in unique_categories:
        output['encoding_map'][unique_category] = category_to_index[unique_category]
    
    for category in categories:       
        if category in category_to_index:
            output['encoded_data'].append(category_to_index[category])
    
    output['num_categories'] = len(unique_categories)
    
    return output

encode_categories = encode_categories(['cat', 'dog', 'cat', 'bird', 'cat', 'dog'])
print(encode_categories)
```

```python
# Problem 8: Dataset Overlap Analyzer

def analyze_overlap(train_ids, test_ids):
    train_set = set(train_ids)
    test_set = set(test_ids)
    
    overlapping_ids = train_set.intersection(test_set)
    
    output = {
        'total_unique': 0,
        'overlap': [],
        'only_train': [],
        'only_test': [],
        'has_leakage': False
        
    }
    
    if len(overlapping_ids) > 0:
        output['has_leakage'] = True
        output['overlap'] = list(overlapping_ids)
    
    output['total_unique'] = len(train_set.union(test_set))
    output['only_train'] = list(train_set.difference(test_set))
    output['only_test'] = list(test_set.difference(train_set))
    
    return output

analyze_overlap = analyze_overlap([1, 2, 3, 4, 5], [4, 5, 6, 7, 8])
print(analyze_overlap)
```

```python
# Problem 9: Text Preprocessing - Stopword Remover

def remove_stopwords(text, stopwords):
    words = text.lower().split()
    filtered_words = [word for word in words if word not in stopwords]        
    unique_filtered_words = set()
    
    for word in filtered_words:
        if word not in unique_filtered_words:            
            unique_filtered_words.add(word)
    
    output = {
        'original_text': list(words),
        'filtered_words': filtered_words,
        'num_removed': len(words) - len(filtered_words),
        'unique_filtered': unique_filtered_words
    }
      
    return output

stopwords = {'the', 'is', 'a', 'an', 'in', 'on', 'at'}
text = "the cat is on the mat mat in the house is happy"
print(remove_stopwords(text, stopwords))
```

```python
# Problem 10: Feature Selection - Correlation Filter

def filter_correlated(features, threshold):
    output = {
        'highly_correlated_pairs': set(),
        'features_to_remove': {},
        'featured_to_keep': {}
    }
    
    correlated_features = []
    seen_features = []
    duplicate_features = set()
    
    for i, (key, value) in enumerate(features.items()):
        if(abs(value) >= threshold):
            output['highly_correlated_pairs'].add(key)
            correlated_features.append(key[0])
            correlated_features.append(key[1])
            
    for pair in correlated_features:
        if pair in seen_features:
            duplicate_features.add(pair)
        else:
            seen_features.append(pair)
    
    output['features_to_remove'] = set(duplicate_features)
    output['featured_to_keep'] = set(seen_features) - set(duplicate_features)
        
    return output

correlations = {
    ('f1', 'f2'): 0.95,
    ('f1', 'f3'): 0.5,
    ('f2', 'f4'): 0.92,
    ('f3', 'f4'): 0.3
}

print(filter_correlated(correlations, threshold=0.9))
```

```python
# Problem 11: Hyperparameter Grid Search
from itertools import product

def generate_grid(params):
    keys = params.keys()
    values = params.values()
    combinations = list(product(*values))
    
    output = []
    for combination in combinations:
        config = dict(zip(keys, combination))
        output.append(config)
    
    return output

params = {
    'learning_rate': [0.01, 0.1],
    'batch_size': [32, 64],
    'epochs': [10, 20]
}

result = generate_grid(params)
print(f"Total combinations: {len(result)}")
print("First 3:", result[:3])
```

```python
# Problem 12: Feature Statistics Calculator

import statistics

def calculate_stats(data):
    output = {}
    
    for(key, values) in data.items():        
        output[key] = {
            'mean': None,
            'median': None,
            'std': None,
            'min': 0,
            'max': 0,
            'q1': 0,
            'q3': 0
        }
        
        output[key]['mean'] = float(statistics.mean(values))
        output[key]['median'] = float(statistics.median(values))
        output[key]['std'] = round(float(statistics.stdev(values) if len(values) > 1 else 0.0), 2)
        output[key]['min'] = min(values)
        output[key]['max'] = max(values)
        output[key]['q1'] = float(statistics.quantiles(values, n=4)[0])
        output[key]['q3'] = float(statistics.quantiles(values, n=4)[2])
            
    return output

data = {
    'age': [25, 30, 35, 40, 45],
    'income': [50000, 60000, 70000, 80000, 90000]
}

print(calculate_stats(data))
```
